### Домашнее задание: 

Напишите Bash скрипт, который выполняет следующие шаги:

1. Создаёт директорию для проекта с именем `fastapi_project`.
2. Переходит в эту директорию.
3. Инициализирует виртуальное окружение Python.
4. Устанавливает необходимые зависимости для FastAPI-приложения из файла `requirements.txt`.
5. Создаёт структуру директорий и файлов для простого FastAPI приложения:
   - Основной файл приложения `main.py`.
   - Папка для статических файлов `static/`.
   - Папка для логов `logs/`.
6. Создаёт файл `requirements.txt` с зависимостями:
   - FastAPI
   - Uvicorn (для запуска приложения)
   - Pydantic (для работы с моделями данных)
7. Запускает сервер FastAPI с помощью Uvicorn.

#### Пошаговые требования:

1. **Создание директории и переход в неё**:
    - Создайте директорию `fastapi_project`.
    - Перейдите в эту директорию и убедитесь, что она была успешно создана.

2. **Создание виртуального окружения**:
    - Используйте команду `python3 -m venv venv` для создания виртуального окружения.
    - Активируйте виртуальное окружение.

3. **Создание файла `requirements.txt`**:
    - В файл `requirements.txt` должны быть добавлены следующие зависимости:
      - `fastapi`
      - `uvicorn`
      - `pydantic`

4. **Установка зависимостей**:
    - Установите зависимости с помощью команды `pip install -r requirements.txt`.

5. **Создание структуры директорий и файлов для приложения**:
    - Создайте следующие каталоги:
      - `static/` — для хранения статических файлов.
      - `logs/` — для хранения логов приложения.
    - Создайте файл `main.py`, в котором будет минимальный код FastAPI приложения.

    Пример кода для `main.py`:
    ```python
    from fastapi import FastAPI

    app = FastAPI()

    @app.get("/")
    def read_root():
        return {"message": "Hello, FastAPI!"}
    ```

6. **Запуск приложения FastAPI**:
    - Запустите сервер FastAPI через `uvicorn` с помощью команды:
    ```bash
    uvicorn main:app --reload
    ```

#### Дополнительные требования:

- В скрипте должны быть проверки на наличие установленных инструментов: Python, pip.
- В случае ошибок (например, отсутствие `pip` или `python3` в системе) скрипт должен завершаться с сообщением о проблеме.
- Если директория `fastapi_project` уже существует, скрипт должен либо очистить её, либо предупредить пользователя и предложить продолжить.

#### Примерный план работы скрипта:

```bash
#!/bin/bash

# Проверка наличия Python 3 и pip
if ! command -v python3 &> /dev/null
then
    echo "Python 3 не установлен. Пожалуйста, установите его."
    exit 1
fi

if ! command -v pip &> /dev/null
then
    echo "pip не установлен. Пожалуйста, установите pip."
    exit 1
fi

# Создание директории для проекта
if [ -d "fastapi_project" ]; then
    echo "Директория fastapi_project уже существует. Очищаем её..."
    rm -rf fastapi_project/*
else
    echo "Создаём директорию fastapi_project..."
    mkdir fastapi_project
fi

cd fastapi_project

# Создание виртуального окружения
echo "Создаём виртуальное окружение..."
python3 -m venv venv

# Активируем виртуальное окружение
source venv/bin/activate

# Создание файла requirements.txt
echo "fastapi" > requirements.txt
echo "uvicorn" >> requirements.txt
echo "pydantic" >> requirements.txt

# Установка зависимостей
echo "Устанавливаем зависимости..."
pip install -r requirements.txt

# Создание директорий для приложения
mkdir -p static logs

# Создание файла main.py с минимальным приложением FastAPI
echo "from fastapi import FastAPI" > main.py
echo "app = FastAPI()" >> main.py
echo "@app.get('/')" >> main.py
echo "def read_root():" >> main.py
echo "    return {'message': 'Hello, FastAPI!'}" >> main.py

# Запуск приложения FastAPI
echo "Запускаем сервер FastAPI..."
uvicorn main:app --reload
```

---

#### Ожидаемые результаты:
- При успешном выполнении скрипта, в директории `fastapi_project` будут созданы все необходимые файлы и каталоги.
- Виртуальное окружение будет создано и активировано.
- Зависимости будут установлены.
- Сервер FastAPI будет успешно запущен, и приложение будет доступно по адресу `http://127.0.0.1:8000/`.
